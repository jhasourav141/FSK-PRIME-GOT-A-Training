# Exercise 3: Prompt Engineering for Video Synthesis

**Tool selected:** Google Veo *(change if you use Runway / Pika / Kling / Luma)*

**Inputs:** `image1.jpg` (source 1), `image2.jpg` (source 2), `blended.jpg` (blend of both) from Exercise 2.

**Exact prompt text:**
> Create a 3-second cinematic video using three reference images in sequence. Start with Image 1 (first source image) for the first second. Smoothly cross-dissolve into Image 2 (second source image) while a gentle morph effect begins. End on the blended image, which appears as the natural fusion of the first two. Maintain consistent colour grading, 16:9 aspect ratio, 24 fps, no text or watermarks.

**Camera / transition instructions:**
- Camera: slow, steady zoom-in (dolly in) of about 5% across the full 3 seconds.
- Transition 1 (0.8 s to 1.5 s): cross-dissolve from Image 1 to Image 2.
- Transition 2 (1.8 s to 2.5 s): morph from Image 2 into the blended image.
- Final 0.5 s: hold on the blended image with a soft fade-out.
- Style: smooth, cinematic, soft lighting.
