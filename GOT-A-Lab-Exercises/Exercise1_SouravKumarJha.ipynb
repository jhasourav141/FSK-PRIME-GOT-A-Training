# Exercise 1: EDA & Automated Reporting with Pandas
**Objective:** generate data, plot, compute row-wise mean, export to Excel.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)  # reproducible results
# Step 1: DataFrame with 4 columns of random numbers (100 rows)
df = pd.DataFrame(np.random.rand(100, 4) * 100, columns=['f1','f2','f3','f4'])
df.head()
```

```python
# Step 2: Scatter plot of f2 vs f3
plt.figure(figsize=(7,5))
plt.scatter(df['f2'], df['f3'], alpha=0.7, color='teal')
plt.xlabel('f2')
plt.ylabel('f3')
plt.title('Scatter Plot: f2 vs f3')
plt.grid(True, alpha=0.3)
plt.savefig('scatter_f2_f3.png', dpi=150)
plt.show()
# Observation: values are uniformly random, so no visible correlation.
```

```python
# Step 3: Row-wise arithmetic mean of f1..f4 stored in f5
df['f5'] = df[['f1','f2','f3','f4']].mean(axis=1)
df.head()
```

```python
# Step 4: Export all five columns to Excel
df.to_excel('Exercise1_output.xlsx', index=False)
print('Saved Exercise1_output.xlsx with columns:', list(df.columns))
```
