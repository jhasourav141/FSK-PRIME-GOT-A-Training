# Exercise 5: Interactive AI Prototyping with Gradio
A simple rule-based chatbot with `gr.ChatInterface`. Run the last cell, open the link, then take a screenshot for submission.

```python
import gradio as gr

def respond(message, history):
    m = message.lower()
    if any(w in m for w in ['hello', 'hi', 'namaste']):
        return 'Namaste! I am a demo assistant for NIC officers. How can I help?'
    if 'nic' in m:
        return 'NIC (National Informatics Centre) provides ICT infrastructure and e-governance solutions to the Government of India.'
    if 'ai' in m or 'artificial intelligence' in m:
        return 'AI lets machines perform tasks that normally need human intelligence, such as language understanding, vision and prediction.'
    return f'You said: "{message}". Try asking about NIC or AI.'

demo = gr.ChatInterface(
    fn=respond,
    title='NIC Officer Demo Chatbot',
    description='Welcome! Type a question or click one of the sample prompts below to test the chatbot.',
    examples=['Hello', 'What is NIC?', 'Explain AI in simple words'],
)
```

```python
demo.launch(share=True)  # opens a local and a public link
```
